```python
# Task 1.1 — Read the CSV file using Python's CSV module
# Step 1: Import the librarie
import csv
import json
```

```python
# Step 2: Read the CSV file
# Open the appointments CSV file
with open("data/appointments.csv", "r", newline="") as file:

    # Read the file using DictReader
    reader = csv.DictReader(file)

    # Print all column names
    print("Column Names:")
    print(reader.fieldnames)

    # Store all rows in a list
    rows = list(reader)


# Print total number of rows
print("\nTotal Number of Rows:", len(rows))
```

```python
# Step 3: Create the parse_row() function
# Function to convert required columns into numeric values
def parse_row(row):

    row["age"] = int(row["age"])
    row["fee_inr"] = int(row["fee_inr"])
    row["prior_no_shows"] = int(row["prior_no_shows"])
    row["distance_km"] = float(row["distance_km"])

    return row
```

```python
# Step 4: Separate accepted and rejected rows
# Create two empty lists
accepted = []
rejected = []


# Check every row
for row in rows:

    try:
        converted_row = parse_row(row)
        accepted.append(converted_row)

    except ValueError:
        rejected.append(row)


# Print the results
print("Accepted Rows:", len(accepted))
print("Rejected Rows:", len(rejected))
```

```python
# Step 5: Save rejected rows in JSON format
# Save rejected rows into a JSON file
with open("rejected_rows.json", "w") as file:

    json.dump(rejected, file, indent=2)


print("Rejected rows saved successfully!")
```

```python
# Step 6: Check your completed Task 1.1
print("Total rows read:", len(rows))
print("Accepted rows:", len(accepted))
print("Rejected rows:", len(rejected))

print("\nFirst rejected row:")
print(rejected[0])
```

```python
# Task 1.2 — Inspect and Clean the Data (Pass 1).
# Step 1: Import Pandas and load the original CSV
import pandas as pd

# Load the original appointments file
df = pd.read_csv("data/appointments.csv")

# Display the first five rows
df.head()
```

```python
# Step 3: Check column information
# Display column names, data types and non-null values
df.info()
```

```python
# Step 4: Find missing values
# Count missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Step 5: Count duplicate rows
# Count duplicate records
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Step 6: Remove duplicate rows
# Remove exact duplicate rows
df = df.drop_duplicates()

# Print remaining rows
print("Rows after removing duplicates:", len(df))
```

```python
# Step 7: Clean booking channel names
# Remove extra spaces and standardize capitalization
df["booking_channel"] = (
    df["booking_channel"].str.strip().str.title()
)

# Display unique booking channels
print("Unique Booking Channels:")
print(df["booking_channel"].unique())

# Count unique values
print(
    "Number of unique booking channels:",
    df["booking_channel"].nunique()
)
```

```python
# Step 8: Convert date columns into datetime format
# Convert appointment date into datetime format
df["appt_date"] = pd.to_datetime(
    df["appt_date"],
    errors="coerce"
)

# Convert booking date into datetime format
df["booked_on"] = pd.to_datetime(df["booked_on"])

# Check the data types
print(df[["booked_on", "appt_date"]].dtypes)
```

```python
# Step 9: Remove rows with missing appointment date or status
# Remove rows where appointment date or status is missing
df = df.dropna(subset=["appt_date", "status"])

# Print the final number of rows
print("Rows after cleaning:", len(df))

# Display the cleaned dataset
df.head()
```

```python
# Task 1.3 — Clean Invalid Values, Fill Missing Values and Handle Outliers
# Step 1: Import NumPy
import numpy as np
```

```python
# Step 2: Find and replace invalid ages
# Count ages below 1 or above 100
invalid_age_count = (
    (df["age"] < 1) | (df["age"] > 100)
).sum()

print("Number of invalid ages:", invalid_age_count)
```

```python
# Replace invalid ages with NaN
df.loc[
    (df["age"] < 1) | (df["age"] > 100),
    "age"
] = np.nan

print("Invalid ages replaced successfully!")
```

```python
# Step 3: Fill missing values using the median
# Count missing age values
print("Missing age values:", df["age"].isnull().sum())
```

```python
# Calculate median values
age_median = df["age"].median()
distance_median = df["distance_km"].median()

print("Median age:", age_median)
print("Median distance:", distance_median)
```

```python
# Fill missing age and distance values with their medians
df["age"] = df["age"].fillna(age_median)
df["distance_km"] = df["distance_km"].fillna(distance_median)

# Convert age into whole numbers
df["age"] = df["age"].round().astype(int)

# Check missing values again
print("Missing values after filling:")
print(df[["age", "distance_km"]].isnull().sum())
```

```python
# Step 4: Detect and cap outliers using the IQR method
# Function to detect and cap outliers
def cap_outliers(column_name):

    # Calculate first and third quartiles
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Count values outside the limits
    capped_count = (
        (df[column_name] < lower_limit) |
        (df[column_name] > upper_limit)
    ).sum()

    # Cap outliers using np.clip()
    df[column_name] = np.clip(
        df[column_name],
        lower_limit,
        upper_limit
    )

    # Display results
    print("\nColumn:", column_name)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of values capped:", capped_count)


# Apply the function to both columns
cap_outliers("distance_km")
cap_outliers("fee_inr")
```

```python
# Step 5: Round distance and fee values
# Round distance to one decimal place
df["distance_km"] = df["distance_km"].round(1)

# Round fee to whole rupees
df["fee_inr"] = df["fee_inr"].round().astype(int)

# Display the final maximum fee
print("Maximum fee after cleaning:", df["fee_inr"].max())

# Display the first five cleaned rows
display(df.head())
```

```python
# Task 1.4 — Create Features and Perform NumPy Analysis
# Step 1: Create the lead_days column
# Calculate the number of days between booking and appointment
df["lead_days"] = (
    df["appt_date"] - df["booked_on"]
).dt.days

# Display the result
print("Lead days created successfully!")

display(df[["booked_on", "appt_date", "lead_days"]].head())
```

```python
print("Minimum lead days:", df["lead_days"].min())
print("Maximum lead days:", df["lead_days"].max())
```

```python
# Step 2: Create the no_show column
# Convert appointment status into numerical values
df["no_show"] = (df["status"] == "No-Show").astype(int)

print("No-show column created successfully!")

display(df[["status", "no_show"]].head())
```

```python
print("Total No-Shows:", df["no_show"].sum())

print(
    "No-Show Percentage:",
    round(df["no_show"].mean() * 100, 1),
    "%"
)
```

```python
# Step 3: Create the NumPy array and calculate column means
# Select the required columns and convert them into a NumPy array
X = df[
    ["age", "distance_km", "fee_inr", "lead_days"]
].to_numpy()

# Calculate the mean of each column
column_means = np.mean(X, axis=0)

print("Mean of each column:")
print(column_means)
```

```python
# Step 4: Find appointments booked more than 14 days ahead
# Create a Boolean mask
mask = df["lead_days"] > 14

# Count appointments booked more than 14 days ahead
count_above_14 = mask.sum()

# Calculate the share of such appointments
share_above_14 = mask.mean()

print("Appointments with lead_days > 14:", count_above_14)

print(
    "Share of appointments:",
    round(share_above_14 * 100, 1),
    "%"
)
```

```python
# Step 5: Calculate distance z-scores
# Convert distance column into a NumPy array
distance = df["distance_km"].to_numpy()

# Calculate z-scores using population standard deviation
z_scores = (
    distance - np.mean(distance)
) / np.std(distance)

# Count values where absolute z-score is greater than 2
outlier_count = (np.abs(z_scores) > 2).sum()

print("Number of distance values with |z| > 2:", outlier_count)
```

```python
# Task 1.5 — Enrich Clinic Data Using the Open-Meteo API
# Step 1: Import the required libraries
import requests
import pandas as pd
import time
```

```python
# Step 2: Load the clinics CSV file
# Load clinic information
clinics = pd.read_csv("data/clinics.csv")

# Display the clinic data
display(clinics)

# Check the number of clinics
print("Total clinics:", len(clinics))
```

```python
# Step 3: Create a function to get temperature
# Function to get temperature from Open-Meteo API
def get_temperature(latitude, longitude):

    url = "https://api.open-meteo.com/v1/forecast"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m",
        "forecast_days": 1
    }

    # Try two times if the request fails
    for attempt in range(2):

        try:
            response = requests.get(
                url,
                params=params,
                timeout=10
            )

            # Check the response status
            if response.status_code != 200:
                raise ValueError(
                    f"Status code: {response.status_code}"
                )

            # Convert response into JSON
            data = response.json()

            # Get hourly temperature information
            temperatures = data["hourly"]["temperature_2m"]

            if not temperatures:
                raise ValueError("Temperature data is empty")

            # Return the first temperature value
            return temperatures[0]

        except (
            requests.exceptions.RequestException,
            ValueError,
            KeyError,
            TypeError
        ) as error:

            reason = str(error)

            if attempt == 0:
                print("Request failed. Retrying...")
                time.sleep(1)

    # Return None if both attempts fail
    print("Could not get temperature:", reason)

    return None
```

```python
# Step 4: Get temperatures for all 8 clinics
# Create an empty list to store temperatures
temperatures = []

# Get temperature for each clinic
for index, row in clinics.iterrows():

    temperature = get_temperature(
        row["latitude"],
        row["longitude"]
    )

    temperatures.append(temperature)

    print(
        row["clinic_name"],
        "Temperature:",
        temperature
    )
```

```python
# Step 5: Add the temperature column
# Add temperature values to the clinics DataFrame
clinics["temp_c"] = temperatures

# Display the updated clinic data
display(clinics)
```

```python
# Step 6: Save the enriched clinic data
# Save the updated clinic data into a CSV file
clinics.to_csv("clinics_enriched.csv", index=False)

print("clinics_enriched.csv saved successfully!")
```

```python
# Step 6: Create a function to format SQL values
import numpy as np

# Function to convert Python values into SQL values
def sql_value(value):

    # Convert missing values into SQL NULL
    if pd.isna(value):
        return "NULL"

    # Numbers do not need quotation marks
    if isinstance(value, (int, np.integer)):
        return str(int(value))

    if isinstance(value, (float, np.floating)):
        return str(float(value))

    # Text values need single quotation marks
    value = str(value).replace("'", "''")

    return f"'{value}'"
```

```python
# Step 7: Generate the load_data.sql file
# Open the SQL file for writing
with open("load_data.sql", "w") as file:

    # First, insert all clinic records
    for _, row in clinics_enriched.iterrows():

        clinic_values = [
            row["clinic_id"],
            row["clinic_name"],
            row["city"],
            row["tier"],
            row["latitude"],
            row["longitude"],
            row["temp_c"]
        ]

        values = ", ".join(
            sql_value(value) for value in clinic_values
        )

        sql_query = f"INSERT INTO clinics VALUES ({values});\n"

        file.write(sql_query)


    # Next, insert all appointment records
    for _, row in appointments_clean.iterrows():

        appointment_values = [
            row["appointment_id"],
            row["patient_id"],
            row["clinic_id"],
            row["booked_on"],
            row["appt_date"],
            row["age"],
            row["gender"],
            row["distance_km"],
            row["fee_inr"],
            row["booking_channel"],
            row["visit_type"],
            row["reminder_sent"],
            row["prior_no_shows"],
            row["status"],
            row["lead_days"],
            row["no_show"]
        ]

        values = ", ".join(
            sql_value(value) for value in appointment_values
        )

        sql_query = f"INSERT INTO appointments VALUES ({values});\n"

        file.write(sql_query)


print("load_data.sql created successfully!")
```
